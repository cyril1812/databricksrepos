```python
source_catalog = "salama_insurance_snowflake_source"
source_db = "salamasilver"
target_catalog = "salama_insurance"
target_schema = "salama_silver"

# Get all tables from federated source
tables = spark.sql(f"""
SHOW TABLES IN {source_catalog}.{source_db}
""").collect()

for row in tables:
    table_name = row['tableName']
    
    source_table = f"{source_catalog}.{source_db}.{table_name}"
    target_table = f"{target_catalog}.{target_schema}.{table_name}"
    
    print(f"Copying {source_table} → {target_table}")
    
    spark.sql(f"""
    CREATE OR REPLACE TABLE {target_table} AS
    SELECT * FROM {source_table}
    """)
```
